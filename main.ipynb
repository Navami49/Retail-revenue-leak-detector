## Retail Revenue Leak Detector
## Project Overview
This project analyzes retail sales data to identify products and business areas that are losing money.
I used Python for data analysis, PostgreSQL to investigate business questions using SQL, and Tableau to create an interactive dashboard.
The main goal is to understand what affects profitability and help businesses make better decisions about discounts, products, and regional performance.

## Tools Used
- Python
- Pandas
- PostgreSQL
- Tableau Public

## Key Questions
- How much sales revenue and profit did the business generate?
- Which product sub-categories are making losses?
- How do discounts relate to profitability?
- Which regions perform best and worst?
- Which products have the lowest total profit?

## Dataset
The Sample Superstore dataset contains 9,994 records and 21 columns, including sales, profit, discounts, products, customers, and regions.

## Conclusion

The analysis showed that the business generated approximately $2.29 million in sales and $286,397 in profit.

Tables, Bookcases, and Supplies were the sub-categories with negative total profit. The analysis also showed that some higher discount levels were associated with lower average profit.

The West region generated the highest total profit, while the Central region generated the lowest.

Based on these findings, the business should review loss-making products, examine discount policies, and investigate ways to improve profitability in weaker regions.
